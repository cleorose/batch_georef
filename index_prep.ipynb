### Adding fields to main tracker table

columns to add: 

- download_filename
- output_filename

- geometry_utm (GeoSeries in EPSG:26915)
- ml_confidence
- gcp_count
- error_log

#### filenames

```python
# create a safe file name column
gdf["download_filename"] = (
    gdf["filename"]
    .astype(str)
    .str.lstrip("/")
    .str.replace("/", "_")
)
```

```python
# create a safe output filename column
gdf["output_filename"] = (
    gdf["download_filename"].apply(lambda f: f.replace(".jpg", "_georef.tif"))
)
```

```python
gdf["processed"] = False
```
